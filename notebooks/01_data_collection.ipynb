# 장보고기지, 세종기지 주변 온도, 습도, 풍속 데이터 수집
## 데이터 출처: Open-Meteo (Historical Weather API)

```python
import requests
import pandas as pd

def fetch_polar_weather_full(url, station_name):
    response = requests.get(url).json()
    
    # 기상청 데이터와 1:1 매칭되도록 5개 지표 모두 추출
    df = pd.DataFrame({
        'Date': response['daily']['time'],
        'Temp_Mean': response['daily']['temperature_2m_mean'],
        'Temp_Max': response['daily']['temperature_2m_max'],
        'Temp_Min': response['daily']['temperature_2m_min'],
        'Humidity_Mean': response['daily']['relative_humidity_2m_mean'],
        'Wind_Mean': response['daily']['wind_speed_10m_mean']
    })
    
    file_path = f'../data/openmetro_{station_name}.csv'
    df.to_csv(file_path, index=False)
    print(f"✅ {station_name.upper()} 기지 데이터(기온3종, 습도, 풍속 완비) 저장 완료: {file_path}")

# 세종기지 & 장보고기지 URL (최고/최저 기온, 평균 풍속으로 파라미터 수정)
sejong_url = "https://archive-api.open-meteo.com/v1/archive?latitude=-62.22&longitude=-58.66&start_date=2016-01-01&end_date=2025-12-31&daily=temperature_2m_mean,temperature_2m_max,temperature_2m_min,relative_humidity_2m_mean,wind_speed_10m_mean&timezone=auto"
jangbogo_url = "https://archive-api.open-meteo.com/v1/archive?latitude=-74.62&longitude=164.22&start_date=2016-01-01&end_date=2025-12-31&daily=temperature_2m_mean,temperature_2m_max,temperature_2m_min,relative_humidity_2m_mean,wind_speed_10m_mean&timezone=auto"

fetch_polar_weather_full(sejong_url, 'sejong')
fetch_polar_weather_full(jangbogo_url, 'jangbogo')
```
